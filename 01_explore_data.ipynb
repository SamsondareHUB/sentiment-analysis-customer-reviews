```python
fix code
# Detect likely text and rating columns more robustly
text_column = None
rating_column = None

# Look for a text-like column: object/string dtype with longer text values
for column in df.columns:
    series = df[column].dropna()
    if series.empty:
        continue
    if pd.api.types.is_object_dtype(series) or pd.api.types.is_string_dtype(series):
        text_values = series.astype(str).str.strip()
        text_values = text_values[text_values != ""]
        if not text_values.empty and text_values.str.len().mean() > 10:
            text_column = column
            break

# Look for a rating-like column: numeric values with a small number of distinct values
for column in df.columns:
    numeric_series = pd.to_numeric(df[column], errors="coerce").dropna()
    if numeric_series.empty:
        continue
    unique_vals = numeric_series.unique()
    if 1 <= len(unique_vals) <= 10 and numeric_series.between(0, 10).all():
        rating_column = column
        break

print("Likely text column:", text_column)
print("Likely rating column:", rating_column)

if text_column is not None:
    print("\nSample review:")
    print(df[text_column].dropna().iloc[0])

if rating_column is not None:
    print("\nUnique ratings:")
    print(sorted(pd.to_numeric(df[rating_column], errors="coerce").dropna().unique()))
```
